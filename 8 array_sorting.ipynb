## array Sorting

```python
import numpy as np

unsorted= np.array([2,4,7,5,1,2,8,9,0])
print("Sorted array: ", np.sort(unsorted))

# 2d array
unsorted_2d= np.array([[4,1], [9,5], [6,2]])
print("Sorted 2D array by column: ", np.sort(unsorted_2d, axis=0)) #column wise sort
print("Sorted 2D array by row: ", np.sort(unsorted_2d, axis=1)) #row wise sort

```

## Advance Operation


```python
import numpy as np
import matplotlib.pyplot as plt

# data strucutre: [restauren_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # kacchi vai
    [2, 120000, 140000, 160000, 190000],  # kacchi dine
    [3, 200000, 230000, 260000, 300000],  # sultan dine's
    [4, 180000, 210000, 240000, 270000],  # chillox
    [5, 160000, 185000, 205000, 230000],  # pizzaburg
])

print("=== FOODPANDA ===")

print("\nSales data Shape:", sales_data.shape)

print("\nData for the 1st 3 restaurent:\n", sales_data[0:3])

# : -> ALL rows && 1: -> start from column index 1 until the end.
print("\nData for the 1st 3 restaurent without indexing:\n", sales_data[:, 1:])

print("\nTotal sales per year:")
yearly_sale= np.sum(sales_data[:, 1:], axis=0)
print(yearly_sale)

print("\nMinimun sales per restaurant:")
min_sales= np.min(sales_data[:, 1:], axis=1)
print(min_sales)

print("\nMaximun sales per year:")
max_sales= np.max(sales_data[:, 1:], axis=0)
print(max_sales)

print("\nAverage sales per restaurant:")
avg_sales= np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

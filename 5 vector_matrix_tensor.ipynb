## Vector, Matrix, Tensor

```python
import numpy as np
```

```python
vector= np.array([1,2,3])
print("Vector: ", vector)

# 2d array
matrix= np.array([[1,2,3],
                 [4,5,6]])

print("Matrix: ", matrix)

# more / complex dimension 
tensor= np.array([[[1,2], [3,4]],
                  [[5,6], [7,8]]])
print("Tensor: ",tensor)
```
